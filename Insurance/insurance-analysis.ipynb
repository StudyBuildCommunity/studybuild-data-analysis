```python
#Q5
pareto = df1[['IDpol','ClaimAmount']].sort_values(
    by='ClaimAmount', ascending=False).reset_index(drop=True)
pareto['cumulative_amount']= pareto['ClaimAmount'].cumsum()
pareto['cumulative_pct']= (pareto['cumulative_amount']/pareto['ClaimAmount'].sum()*100)
pareto['claim_pct']= ((pareto.index+1)/len(pareto)*100)
pareto
```

```python
# the largest 20% of claims account for approximately 69% of the total claim amount.
n = int(len(pareto)*0.20)
top_20_pct= (pareto.iloc[:n]
             ['ClaimAmount'].sum()/pareto['ClaimAmount'].sum()*100)
top_20_pct
```

```python
plt.plot(pareto['claim_pct'],pareto['cumulative_pct'])
plt.xlabel('cumulative % of claims')
plt.ylabel('cumulative % of claim cost')
plt.title('pareto analysis: claims vs cost')
plt.grid()
plt.show()
```

```python
driver_age = df2.groupby('DriverAgeGroup').agg( claims=('ClaimNb', 'sum'), exposure=('Exposure', 'sum'), claim_amount=('ClaimAmount', 'sum'))
driver_age['frequency']=driver_age['claims']/driver_age['exposure']
driver_age['severity']=driver_age['claim_amount']/driver_age['claims']
driver_age [driver_age['claims']>0]
```

```python
plt.scatter(driver_age['frequency'], driver_age['severity'])
plt.xlabel('Claim Frequency')
plt.ylabel('Claim Severity')
plt.title('Claim Frequency vs Severity by Driver Age')
plt.show()
```
